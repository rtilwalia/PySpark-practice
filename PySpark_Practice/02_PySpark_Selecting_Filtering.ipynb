```python
from pyspark.sql import SparkSession
df = spark.read.table("samples.bakehouse.sales_transactions")
```

```python
df.dtypes
```

## Hands-on Challenge

### Part A: Selecting
- Select only 4 columns.
- Rename one column using alias().
- Create a new column by multiplying a numeric column by 2.

```python
from pyspark.sql.functions import col
```

```python
df_select = df.select(
                col("product"),
                col("quantity"),
                col("totalPrice").alias("final_price"),
                (col("unitPrice")*2).alias("double_price")
)

df_select.show(10)
```


### Part B: Filtering


```python
#Values greater than a number

df_select.filter(col("quantity") > 10).show()
```

```python
# Values less than a number.

df_select.filter(col("quantity") < 5).show()
```

```python
#Two conditions using |

df_select.filter((col("product") == "Tokyo Tidbits") | (col("product") == "Golden Gate Ginger")).show()
```

```python
# Two conditions using &

df_select.filter((col("product") == "Tokyo Tidbits") & (col("product") == "Golden Gate Ginger")).show()
```

```python
# A between() filter

df_select.filter(col("final_price").between(10,25)).show()
```

```python
#isin() filter

df_select.filter(col("product").isin("iPhone 11 Pro Max", "iPhone 11 Pro")).show()
```

```python
#Rows where a chosen column is NULL
df_select.filter(col("final_price").isNull()).show()
```

```python
#Rows where a chosen column is NULL
df_select.filter(col("final_price").isNotNull()).show()
```


- Why is col() preferred over df.columnName in production code?
1.  Catches better IDE support and better for error handling
2. Works with any column name, even those containing spaces or reserved words.
3. Makes chained expressions more readable and consistent.
- Why do we use & instead of and in PySpark filters?
1. and, or, is not supported by python
2. col("quantity") > 10 returns a Spark Column expression, not  a Python bool. Column<'(quantity > 10)'>

3. Python's and expects True or False, but Spark overloads the bitwise operators (&, |, ~) to build expression trees.

- Are filter() and where() different internally?
1. No
