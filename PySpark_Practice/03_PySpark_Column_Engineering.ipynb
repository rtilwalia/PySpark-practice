```python
from pyspark.sql import SparkSession
df = spark.read.table("samples.bakehouse.sales_transactions")
```

```python
from pyspark.sql.functions import col
```

### withColumn

eg -
df.withColumn(
    "country",
    lit("India")
)

* Functions used with withColumn
- lit()
- cast()
- when()/otherwise()
- upper()
- lower()
- initcap()
- trim()

```python
df.show(10)
```

```python
df_discount = df.withColumn("discount_price", col("totalPrice")*0.9)
df_discount.show(10)
```

```python
from pyspark.sql.functions import lit

df_country = df_discount.withColumn("country", lit("India"))
df_country.show(10)
```

```python
from pyspark.sql.functions import when

df_order = df_country.withColumn(
    "large_order",
    when(col("quantity") >= 20, "YES")
    .otherwise("NO")
)

df_order.show(10)
```

```python
from pyspark.sql.functions import *

df_upper = df_order.withColumn("product_upper", upper(col("product")))
df_upper.select(col("product_upper")).show(10)
```

```python
df_lower = df_order.withColumn("product_lower", lower(col("product")))
df_lower.select(col("product_lower")).show(10)
```

```python
df_initcap = df_order.withColumn("product_initcap", initcap(col("product")))
df_initcap.select(col("product_initcap")).show(10)
```

```python
df_trim = df_order.withColumn("product_trim", trim(col("product")))
df_trim.select(col("product_trim")).show(10)
```

```python
df.dtypes
```

```python
df_double  = df_order.withColumn("quantity", col("quantity").cast("double"))

df_double.dtypes
```

### Bonus Challenge
I/P - 
| quantity |
| -------: |
|        5 |
|       12 |
|       18 |
|       25 |

O/P - 
| quantity | category |
| -------: | -------- |
|        5 | Small    |
|       12 | Medium   |
|       18 | Medium   |
|       25 | Large    |


```python
df_ans = df_order.withColumn(
    "Category",
    when(col("quantity") <= 5, "Small")
    .when((col("quantity") > 5) & (col("quantity") <= 20), "Medium")
    .otherwise("Large")
)

df_ans.show(10)
```

- Today's seed:

    Why do we create Bronze → Silver → Gold instead of one giant ETL job?

    We'll answer that properly when we reach Delta Lake, but start thinking about it now. Every design decision in data engineering has trade-offs.

> My Answer
    - All three solves different purpose, 
    - Bronze - Contains raw data
    - Silver - Transformation logic is applied as per individual team's requirement
    - Gold - Serves the user with different story of data
    - Three individual layers mean that they can serve different requirement and if data is changed due to business requirement or source changes it can be handled easily whereas handling this in one giant ETL job can be cumbersome
