day 1 challenge

What is the difference between printSchema() and show()? printschema only prints the schema and types of the columns and show prints out the no of rows mentioned
Which of the commands above are actions? - all were actions except len
Which are transformations? - len
Why does df.count() usually take longer than df.show(5)?


| Command            | Action? | Reason                                                      |
| ------------------ | ------- | ----------------------------------------------------------- |
| `df.printSchema()` | ❌ No    | Uses schema metadata already available.                     |
| `df.show(5)`       | ✅ Yes   | Triggers Spark to retrieve and display rows.                |
| `df.columns`       | ❌ No    | Returns column names from metadata.                         |
| `len(df.columns)`  | ❌ No    | Pure Python operating on the list returned by `df.columns`. |
| `df.count()`       | ✅ Yes   | Must scan the dataset to count rows.                        |


but none of them are transformations either
